```python
import numpy as np
import matplotlib.pyplot as plt
from lab_utils_multi import zscore_normalize_features, run_gradient_descent_feng
np.set_printoptions(precision=2)
```

```python
x = np.arange(0,20,1)
y = 1 + x**2
X = x.reshape(-1,1)
```

```python
model_w, model_b = run_gradient_descent_feng(X, y, iterations = 1000, alpha = 1e-2)
```

```python
plt.scatter(x, y, marker='x', c = 'r', label = 'Actual Value')
plt.title("no feature engineering")

plt.plot(x, X@model_w + model_b, label = 'Predicted Values')
plt.xlabel('X')
plt.ylabel('y')

plt.legend()
plt.show()
```

```python
# Raw Data
x = np.arange(0,20,1)

# Equation
y = 1 + x**2

# Engineered Feature
X = x**2
```

```python
X = X.reshape(-1, 1)
model_w, model_b = run_gradient_descent_feng(X, y, iterations = 1000, alpha = 1e-5)
```

```python
plt.scatter(x, y, marker = 'x', c = 'r', label = 'Actual Value')
plt.title('Added x**2 features')

plt.plot(x, np.dot(X,model_w) + model_b, label = 'Predicted Values');
plt.xlabel('X')
plt.ylabel('y')

plt.legend()
plt.show()
```

```python
# Creating Dataset

x = np.arange(0,20,1)
y = x**2

# Engineered Features
X = np.c_[x, x**2, x**3]
```

```python
model_w, model_b = run_gradient_descent_feng(X, y, iterations = 10000, alpha = 1e-7)
```

```python
plt.scatter(x, y, marker = 'x', c = 'r', label = 'Actual Value')
plt.title('x, x**2, x**3')

plt.plot(x, np.dot(X, model_w) + model_b, label = 'Predicted Values')
plt.xlabel('x')
plt.ylabel('y')

plt.legend()
plt.show()
```

```python
# Create Target Data
x = np.arange(0,20,1)
y = x**3

# Create Engineered Features
X = np.c_[x, x**2, x**3]
```

```python
# After Traning
model_w, model_b = run_gradient_descent_feng(X, y, iterations = 10000, alpha = 1e-8)
```

```python
# Here weight of x**3 is highest and x is 0
plt.scatter(x, y, marker = 'x', c = 'r', label = 'Actual Value')
plt.title('x, x**2, x**3')

plt.plot(x, np.dot(X, model_w) + model_b, label = 'Predicted Values')
plt.xlabel('x')
plt.ylabel('y')

plt.legend()
plt.show()
```
