### What are disadvantages of One Hot Encoding ?
1.  Curse of Dimensionality:

    Creates high-dimensional sparse matrices

    Each unique category adds a new column to the dataset

    As categories increase, the number of dimensions grows exponentially

2. Memory and Computational Issues:

    Requires more memory to store the transformed data

    Increases computational complexity

    Can slow down model training, especially with limited resources

    Creates sparse matrices where most values are zeros

3. Loss of Information:

    Doesn't capture relationships between categories

    No semantic meaning between related categories

    For example, words like "run", "walk", and "shoe" would have no relationship in the encoded space despite being semantically related

4. Potential Overfitting:

    The increased number of features can lead to overfitting

    Models might learn patterns in the sparse data that don't generalize well

5. Scalability Issues:

    Not practical for categorical variables with many unique values

    For example, if you have a category with 1000 unique values, you'll end up with 1000 new columns

6. Storage Inefficiency:

    Requires more storage space

    Most values in the matrix will be zeros, leading to inefficient storage utilization



```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
```

```python
data['color']
```

```python
encoded_data = encoder.fit_transform(data[['color']])
```

```python
encoded_data
```

```python
encoded_df = pd.DataFrame(encoded_data, columns=encoder.get_feature_names_out(['color']))
```

```python
encoded_df
```

```python
final_df = pd.concat([data, encoded_df], axis=1)
```
