
    Import nltk: This line imports the nltk library, making its functions available.
    Download punkt_tab: This line uses nltk.download('punkt_tab') to download the necessary Punkt Tokenizer models. This will store the models in your NLTK data directory, so you only need to download them once.
    Import sent_tokenize: Imports the function needed for tokenization.
    Tokenize and Print: This line calls sent_tokenize(corpus) to tokenize the text stored in the corpus variable, and then it prints the resulting list of sentences.


```python
## Tokenisation

## Paragraphs ---> Sentences

# Import Library
import nltk

# Download Punkt Tokenizer Model
nltk.download('punkt_tab')

# Import sent_tokenize function to tokenize the corpus
from nltk.tokenize import sent_tokenize
```

```python
documents = sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
  print(sentence)
```

```python
from nltk.tokenize import word_tokenize
```

```python
vocabulary = word_tokenize(corpus)
```

```python
for word in vocabulary:
  print(word)
```

```python
# in this case the punctuation is considered as a different sentence

wordpunct_tokenize(corpus)
```
