# Lemmatizer

In Natural Language Processing (NLP), Lemmatization is the process of reducing a word to its base or dictionary form, known as the lemma. It considers the context and meaning of the word to ensure the output is a valid word in the language.

How is Lemmatization different from Stemming?

While both stemming and lemmatization aim to reduce words to their root forms, Lemmatization is more sophisticated. Stemming simply removes common suffixes, which can sometimes result in non-words. Lemmatization, on the other hand, uses vocabulary analysis and morphological knowledge to identify the lemma.

Why is Lemmatization used?

Lemmatization is useful for various NLP tasks, including:

    Information Retrieval: Improves search accuracy by grouping different forms of a word together.
    Text Classification: Reduces the dimensionality of the data and improves model performance.
    Machine Translation: Helps in mapping words across languages based on their core meaning.
    Chatbots and Conversational AI: Enables better understanding of user input.

Example

The word "better" would be lemmatized to "good" because it is the root form of the adjective. Stemming, on the other hand, might just remove the "er" suffix, leaving "bett," which is not a word.

# WordNetLemmatizer

WordNetLemmatizer is a specific lemmatizer available within the NLTK library in Python. NLTK stands for Natural Language Toolkit, and it's a popular library for working with human language data.

WordNetLemmatizer utilizes WordNet, a large lexical database of English. WordNet groups words into sets of synonyms called synsets and provides relationships between these synsets.

How it Works

    Input: You provide a word to the WordNetLemmatizer.
    Lookup: It looks up the word in WordNet to identify its possible lemmas (base forms).
    Context: If necessary, it considers the part-of-speech (POS) tag of the word to refine the lemma selection.
    Output: It returns the most appropriate lemma for the word based on WordNet's information.


In this example, by specifying pos='v', we tell the lemmatizer to consider "running" as a verb, leading to the correct lemma "run." Without the POS tag, it might have lemmatized it to "running" itself (treating it as a noun).

Why use WordNetLemmatizer?

    Accuracy: It uses a comprehensive lexical resource (WordNet) for lemmatization, resulting in high accuracy.
    Contextual Awareness: It can consider POS tags to further refine the lemma selection.
    Ease of Use: NLTK provides a simple interface to use WordNetLemmatizer.

```python
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
corpus = """Children were running around the cats while singing loudly in their house. They had been dancing all day, which made them feel exhausted but happy. The more they jumped with joy,
the more enjoyment they found in every moment. Although they couldn't remember exactly how many times they had stolen candies from the store, it was a fun memory. Meanwhile, workers
were building new houses nearby, and neighbors often played together after work. Some people walked to the park, while others talked about their plans for the weekend."""

# Example words to compare stemming vs lemmatization:
special_cases = [
    "am", "is", "are", "was", "were",  # be verbs
    "better", "best", "good",          # irregular adjectives
    "mice", "mouse",                   # irregular plurals
    "walking", "walked", "walks",      # regular variations
    "studies", "studying", "studied",  # y->i changes
    "analyses", "analyzing", "analyzed" # special spellings
]
```

```python
lemmatizer.lemmatize('going')
```

```python
'''
POS
Noun - n
verb - v
adjective - a
adverb - r
'''
```

```python
lemmatizer.lemmatize('going', pos = 'n')
```

```python
lemmatizer.lemmatize('going', pos = 'v')
```

```python
lemmatizer.lemmatize('going', pos = 'a')
```

```python
lemmatizer.lemmatize('going', pos = 'r')
```

# Tokenization ---> Stemming ---> Lemmatization

### Tokenisation

```python
# Download punkt Tokenizer Model
nltk.download('punkt_tab')

# Import word_tokenize function to tokenize the corpus
from nltk.tokenize import word_tokenize
```

```python
words = word_tokenize(corpus)
```

### Stemming

```python
import pandas as pd

# Creating a column with original words
df = pd.DataFrame({'original_words': words})

# Initializing the stemmers
porter = nltk.PorterStemmer()
regexp = nltk.RegexpStemmer('ing$|s$|e$|able$', min=4)
snowball = nltk.SnowballStemmer(language='english')

# Doing Stemmeing
df['porter'] = df['original_words'].apply(lambda x : porter.stem(x) )
df['regexp'] = df['original_words'].apply(lambda x : regexp.stem(x) )
df['snowball'] = df['original_words'].apply(lambda x : snowball.stem(x) )

# Doing lemmatization
df['lemmatization_n'] = df['original_words'].apply(lambda x : lemmatizer.lemmatize(x, pos = 'n'))
df['lemmatization_v'] = df['original_words'].apply(lambda x : lemmatizer.lemmatize(x, pos = 'v'))
df['lemmatization_a'] = df['original_words'].apply(lambda x : lemmatizer.lemmatize(x, pos = 'a'))
df['lemmatization_r'] = df['original_words'].apply(lambda x : lemmatizer.lemmatize(x, pos = 'r'))

# Display
df
```
