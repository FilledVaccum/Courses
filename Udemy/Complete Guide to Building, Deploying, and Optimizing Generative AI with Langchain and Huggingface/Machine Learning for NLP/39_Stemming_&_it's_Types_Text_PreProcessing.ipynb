# Stemming

What is Stemming?

Stemming is a text preprocessing technique used in Natural Language Processing (NLP) to reduce words to their root or base form, called a stem.

Example:

The words "playing", "played", and "plays" would all be reduced to the stem "play".

Why is it used?

    Reduces dimensionality: Stemming reduces the number of unique words in a text, which can improve the performance of machine learning models.
    Improves information retrieval: By grouping similar words together, stemming can help search engines and other information retrieval systems find relevant documents more effectively.
    Simplifies text analysis: Stemming can make it easier to analyze text data by reducing the complexity of the vocabulary.

Things to keep in mind:

    Stemming can sometimes produce stems that are not real words.
    There are different stemming algorithms available, each with its own strengths and weaknesses. Choose the one that best suits your needs.
    Stemming is just one of many text preprocessing techniques. Consider using it in conjunction with other techniques like tokenization, stop word removal, and lemmatization for better results.

# Porter Stemmer

The Porter Stemmer is a widely used algorithm for stemming in natural language processing (NLP). It was developed by Martin Porter in 1980 and is known for its simplicity and effectiveness.

How it works:

The Porter Stemmer is a rule-based stemmer. It applies a series of rules to words to remove suffixes and reduce them to their base form. The rules are based on common morphological patterns in English.

Algorithm:

The Porter Stemmer algorithm involves five phases of word reductions, applied sequentially. Each phase consists of a set of rules. A word is passed through each phase, and if a rule applies, the suffix is removed, and the process continues to the next phase. The algorithm stops when no more rules apply.

Example:

Consider the word "running." Here's how the Porter Stemmer would process it:

    Phase 1: The suffix "ing" is identified and removed, resulting in "run."
    Phase 2: No rules apply to "run" in this phase.
    Phase 3: No rules apply to "run" in this phase.
    Phase 4: No rules apply to "run" in this phase.
    Phase 5: No rules apply to "run" in this phase.

Therefore, the stemmed word for "running" is "run."

Advantages:

    Simple to understand and implement.
    Relatively fast and efficient.
    Widely used and well-tested.

Disadvantages:

    Can sometimes over-stem or under-stem words.
    May not be suitable for all languages or domains.
    Rule-based approach may not capture all morphological variations.

In summary, the Porter Stemmer is a popular and effective algorithm for stemming in English. It is a good choice for many NLP tasks, but it's essential to be aware of its limitations.

# RegexpStemmer

The RegexpStemmer class is a stemming algorithm that uses regular expressions to identify and remove morphological affixes from words. It's part of the nltk.stem module in the NLTK library, a popular Python library for natural language processing.

How it works:

    You provide a regular expression pattern to the RegexpStemmer constructor. This pattern defines the affixes you want to remove.
    When you call the stem method on a word, the RegexpStemmer tries to match the pattern against the word.
    If a match is found, the affix is removed, and the stemmed word is returned.


```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min = 4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingablecodes')
```

```python
reg_stemmer = RegexpStemmer('ing|s$|e$|able$', min = 4)
```

```python
reg_stemmer.stem('ingcodeing')
```

```python
# Create a RegexpStemmer with a pattern to remove 'ing', 's', and 'ed' suffixes
st = RegexpStemmer('ing$|s$|ed$', min=4)

# Stem some words
for word in words:
    stemmed_word = st.stem(word)
    print(f"Original word: {word}, Stemmed word: {stemmed_word}")
```

Explanation:

    'ing$|s$|ed$': This regular expression pattern matches words ending with 'ing', 's', or 'ed'.
    min=4: This optional argument specifies the minimum length of the word after stemming. In this case, words shorter than 4 characters after stemming will not be modified.

Advantages:

    Flexibility: You can define your own stemming rules using regular expressions, making it adaptable to different languages and specific needs.
    Speed: Regular expression matching is generally fast, making RegexpStemmer efficient for large datasets.

Disadvantages:

    Complexity: Creating effective regular expressions for stemming can be challenging and require expertise.
    Over-stemming or under-stemming: If the regular expressions are not carefully designed, they might remove too much or too little from words, leading to inaccurate stemming.

When to use it:

    When you need more control over the stemming process than provided by standard stemming algorithms.
    When you have specific affixes you want to remove.
    When you need a fast stemming algorithm for large datasets.


# Snowball Stemmer
The Snowball Stemmer, also known as the Porter2 Stemmer, is an improved version of the original Porter Stemmer. It addresses some of the limitations of the Porter Stemmer and provides support for multiple languages.

How it works:

Similar to the Porter Stemmer, the Snowball Stemmer is a rule-based stemmer. It applies a series of rules to words to remove suffixes and reduce them to their base form. However, the rules in the Snowball Stemmer are more refined and language-specific.

Algorithm:

The Snowball Stemmer algorithm is based on a language-specific set of rules. It involves multiple steps of suffix removal and replacement, applied sequentially. The algorithm stops when no more rules apply.

Supported Languages:

The Snowball Stemmer supports a wide range of languages, including:

    Arabic
    Danish
    Dutch
    English
    Finnish
    French
    German
    Hungarian
    Italian
    Norwegian
    Portuguese
    Romanian
    Russian
    Spanish
    Swedish

Advantages:

    Improved accuracy compared to the Porter Stemmer.
    Supports multiple languages.
    Relatively fast and efficient.

Disadvantages:

    Can still over-stem or under-stem words in some cases.
    May not be suitable for all domains.
    Rule-based approach may not capture all morphological variations.

In summary, the Snowball Stemmer is a powerful and versatile stemming algorithm that provides improved accuracy and language support compared to the Porter Stemmer. It is a good choice for many NLP tasks, especially those involving multiple languages.

```python
from nltk.stem import SnowballStemmer
```

```python
snowball_obj = SnowballStemmer('english')
```

```python
snowball_obj.stem('eating')
```

```python
snowball_obj.stem('eatable')
```

```python
for word in words:
  print(word+"--->"+snowball_obj.stem(word))
```
