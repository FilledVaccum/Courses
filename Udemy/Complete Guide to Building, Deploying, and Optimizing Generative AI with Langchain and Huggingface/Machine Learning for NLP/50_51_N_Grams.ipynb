## What is N-Gram?


In natural language processing, an N-gram is a contiguous sequence of n items from a given sample of text or speech. These items can be phonemes, syllables, letters, words, or base pairs according to the application. The n in N-gram represents the number of items in the sequence.

Example: Let's consider the sentence: "The quick brown fox jumps over the lazy dog."

Here are a few examples of N-grams from this sentence:

    Unigram (n=1): "The", "quick", "brown", "fox", "jumps", "over", "the", "lazy", "dog"
    Bigram (n=2): "The quick", "quick brown", "brown fox", "fox jumps", "jumps over", "over the", "the lazy", "lazy dog"
    Trigram (n=3): "The quick brown", "quick brown fox", "brown fox jumps", "fox jumps over", "jumps over the", "over the lazy", "the lazy dog"

Uses of N-grams:

N-grams are used in various text analysis and natural language processing tasks such as:

    Text prediction: Predicting the next word in a sequence based on the preceding words.
    Spelling correction: Identifying and correcting spelling errors.
    Machine translation: Translating text from one language to another.
    Sentiment analysis: Determining the sentiment expressed in a piece of text.
    Text summarization: Creating a condensed version of a longer text.
    Document classification: Categorizing documents into predefined classes.
    Authorship attribution: Identifying the author of a text based on their writing style.
    Spam detection: Identifying spam emails or messages.
    Part-of-speech tagging: Identifying the part of speech of each word in a sentence.

## What problem does this concept solve?
Problems solved by N-grams:

N-grams address several challenges in natural language processing and text analysis by providing a way to capture and represent the sequential relationships between words or other linguistic units. Here are some of the key problems they solve:

    Lack of Word Order Information: Traditional "bag-of-words" models treat words as independent entities, ignoring their order. This can lead to inaccurate or incomplete understanding of text meaning. N-grams, especially those with n>1, preserve word order and provide a more accurate representation of text structure.

    Limited Contextual Understanding: Bag-of-words models lack contextual information about the words in a document. N-grams, by considering the surrounding words, capture contextual cues and improve the understanding of word meanings and relationships. This is crucial for tasks like sentiment analysis and text classification.

    Difficulty in Language Modeling: Building statistical language models, which estimate the probability of a sequence of words, is challenging without considering word order and context. N-grams provide the foundation for building such models, enabling applications like speech recognition and machine translation.

    Inefficient Text Representation: Representing text as a simple collection of words can be inefficient for many tasks. N-grams offer a more structured and informative representation, allowing for better feature engineering and improved performance in machine learning models.

    Limited Understanding of Writing Style: Identifying the author of a text or detecting plagiarism is difficult without considering the characteristic patterns of language use. N-grams can capture these patterns, enabling authorship attribution and stylistic analysis.

In summary, N-grams solve the following problems:

    Capturing word order and context
    Improving language modeling
    Providing a more efficient text representation
    Enabling stylistic analysis and authorship attribution
    Enhancing various NLP tasks such as text prediction, sentiment analysis, and machine translation


## How does this concept compare to similar ones?

Comparison of N-grams to similar concepts:

N-grams are often compared to other techniques used for text representation and analysis, such as:

    Bag-of-Words (BoW): BoW models represent text as a collection of individual words, ignoring their order and context. While simple, BoW models lack the ability to capture the relationships between words, which N-grams address by considering word sequences.

    Word Embeddings (Word2Vec, GloVe): Word embeddings represent words as dense vectors in a continuous space, capturing semantic relationships between words. While embeddings capture meaning and similarity, they typically don't explicitly represent word order like N-grams do.

    Skip-grams: Skip-grams are a type of word embedding model that learns to predict context words given a target word. They can capture relationships between words that are not immediately adjacent, but they are primarily focused on word-level semantics rather than sequential patterns.

    Syntactic Parsing: Syntactic parsing involves analyzing the grammatical structure of a sentence, identifying relationships between words based on their syntactic roles. While parsing provides deeper linguistic information, it can be more complex and computationally intensive than N-gram analysis.

    Latent Semantic Analysis (LSA): LSA is a technique for discovering latent semantic relationships between words and documents using matrix factorization. It can capture broader semantic themes but may not be as effective in capturing local word order like N-grams.

In summary:

    N-grams are a simple and effective way to capture word order and context, but they have limited semantic understanding compared to word embeddings.
    Bag-of-Words is even simpler but ignores word order and context.

The choice of technique depends on the specific NLP task and the desired level of linguistic information. N-grams are often used in combination with other techniques to leverage their strengths and address their limitations.

## What assumptions I may be making about this concept, and are they valid?

Assumptions about N-grams and their validity:

When working with N-grams, there are a few key assumptions that are often made:

    The Markov Assumption: This assumption states that the probability of the next word in a sequence depends only on the preceding n-1 words. In other words, the history beyond the immediate context is ignored. This assumption simplifies the model but may not always hold true in real-world language, where longer-range dependencies can exist.

Validity: The Markov assumption is a simplification, but it is often a reasonable approximation for many NLP tasks. The effectiveness of N-gram models in various applications suggests that it captures a significant portion of the relevant information.

    Stationarity: This assumption implies that the statistical properties of the language do not change over time. This means that the N-gram probabilities estimated from a training corpus will be valid for new data from the same domain.

Validity: Stationarity is often violated in practice, as language evolves over time. However, for relatively stable domains or shorter time periods, the assumption can be reasonable. For tasks involving rapidly changing language, using more recent or domain-specific training data is important.

    Data Sparsity: N-gram models rely on counting the occurrences of word sequences in a corpus. For large values of n or infrequent word combinations, the counts can be very low or even zero. This data sparsity can lead to inaccurate probability estimates.

Validity: Data sparsity is a common issue in N-gram models, especially for higher-order N-grams. Techniques like smoothing and backoff are used to mitigate this problem by estimating probabilities for unseen or rare N-grams based on the frequencies of lower-order N-grams.

    Independence of N-grams: N-gram models typically assume that the occurrences of different N-grams are independent of each other. This simplification allows for efficient computation but may not always be accurate, as there can be dependencies between different word sequences.

Validity: The independence assumption is a simplification, but it is often a reasonable approximation for many tasks. More complex models, such as recurrent neural networks, can capture longer-range dependencies and interactions between N-grams.

In summary, while the assumptions made by N-gram models are simplifications of real-world language, they are often reasonable approximations that allow for efficient computation and effective performance in many NLP tasks. It's important to be aware of these assumptions and their potential limitations when applying N-gram models to specific problems.

## How can this concept be broken down into simpler components?

reaking down N-grams into simpler components:

    Text Corpus: The foundation of N-grams is a text corpus, which is a collection of text documents used for analysis. This corpus can be anything from a single sentence to a massive dataset of books or articles.

    Tokenization: Before creating N-grams, the text corpus needs to be tokenized. This involves breaking down the text into individual units, called tokens, which can be words, characters, or other linguistic elements.

    N-gram Generation: Once the text is tokenized, N-grams are generated by sliding a window of size n over the tokens. Each window captures a sequence of n tokens, forming an N-gram.

    Frequency Counting: After generating the N-grams, their frequencies are counted. This involves tallying the number of times each unique N-gram appears in the corpus.

    Probability Estimation: Based on the frequency counts, probabilities can be estimated for each N-gram. This represents the likelihood of that particular sequence occurring in the language or domain represented by the corpus.

    Applications: These N-gram frequencies and probabilities are then used in various NLP applications, such as text prediction, spelling correction, language modeling, and more.

Example:

Let's consider the sentence: "The quick brown fox jumps over the lazy dog."

    Text Corpus: This sentence is our text corpus.
    Tokenization: The sentence is tokenized into words: ["The", "quick", "brown", "fox", "jumps", "over", "the", "lazy", "dog"].
    N-gram Generation (for bigrams, n=2):
        "The quick"
        "quick brown"
        "brown fox"
        "fox jumps"
        "jumps over"
        "over the"
        "the lazy"
        "lazy dog"
    Frequency Counting: Each bigram is counted based on its occurrences in the corpus. In this case, each bigram appears only once.
    Probability Estimation: Since each bigram appears once, their probabilities would be estimated based on the total number of bigrams (8 in this case).
    Applications: These bigrams and their probabilities could be used for various tasks, such as predicting the next word in a sequence or identifying common phrases in the text.

By breaking down the concept into these simpler components, we can better understand the underlying mechanisms and how N-grams are used in practice.

## How does this concept fit into larger systems or frameworks?

N-grams in Larger Systems and Frameworks:

N-grams are a fundamental building block in many larger natural language processing (NLP) systems and frameworks. They serve as a basic unit for representing and analyzing text, and their applications extend to various areas within NLP. Here's how they fit into some common systems and frameworks:

    Statistical Language Models: N-grams form the core of statistical language models, which are used to estimate the probability of a sequence of words. These models are crucial for tasks like speech recognition, machine translation, and text prediction.

    Information Retrieval: In information retrieval systems, N-grams are used for indexing and searching documents. By representing documents as collections of N-grams, search engines can efficiently match queries to relevant documents based on shared N-gram patterns.

    Text Classification and Clustering: N-grams can be used as features in machine learning models for text classification and clustering. By representing documents as vectors of N-gram frequencies, these models can learn to categorize or group documents based on their linguistic content.

    Sentiment Analysis: N-grams are used in sentiment analysis to identify the emotional tone of a text. By analyzing the frequencies of specific N-grams associated with positive or negative sentiment, these systems can determine the overall sentiment expressed in a document.

    Authorship Attribution: N-grams can be used to identify the author of a text based on their writing style. By analyzing the characteristic patterns of N-gram usage in a text, these systems can compare it to known writing samples of different authors to determine the most likely author.

    Machine Translation: In machine translation systems, N-grams are used to align and translate phrases between different languages. By identifying common N-gram patterns in parallel corpora (texts translated into multiple languages), these systems can learn to translate new text based on similar patterns.

Examples of Systems and Frameworks using N-grams:

    Google Search: Uses N-grams for indexing and searching web pages.
    Spam Filters: Employ N-grams to identify spam emails based on common patterns in spam messages.
    Auto-Complete: Uses N-grams to predict the next word in a sentence or phrase.
    Speech Recognition: Utilizes N-grams to improve the accuracy of speech-to-text conversion.
    Natural Language Understanding (NLU) platforms: Like Dialogflow and Rasa, use N-grams in their intent recognition and entity extraction modules.

In essence, N-grams serve as a foundational element for representing and analyzing text data in various NLP systems and frameworks. They enable these systems to understand the structure, meaning, and context of language, ultimately contributing to the advancement of NLP applications.

### Data Cleaning & PreProcessing

```python
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
messages
```

```python
corpus = []
for i in range(0, len(messages)):
  review = re.sub('[^a-zA-Z]', ' ', str(messages['message'][i]))
  review = review.lower()
  review = review.split()
  review = [ps.stem(word) for word in review if not word in stopwords.words('english')]
  review = ' '.join(review)
  corpus.append(review)
```
