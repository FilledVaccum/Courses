# What is ColumnTransformer?

ColumnTransformer is a tool in scikit-learn (a popular machine learning library in Python) used for data preprocessing. It's particularly useful when you need to apply different transformations to different columns of your dataset. For example, you might want to:

    One-hot encode categorical features.
    Scale or standardize numerical features.
    Apply custom transformations to specific columns.

How does it work?

    Define Transformers: You specify the transformations you want to apply, along with the columns they should be applied to.
    Create ColumnTransformer: You create a ColumnTransformer object, passing in the transformers and column selections.
    Fit and Transform: You use the fit_transform method on your data to apply the transformations and get the transformed data.

Why is it useful in Colab/Jupyter Notebooks?

ColumnTransformer simplifies data preprocessing workflows, making your code more organized and readable. This is particularly beneficial in Colab or Jupyter Notebooks, where you often work with data transformations interactively. It also ensures that the same transformations are applied consistently during training and prediction.

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Sample data
data = {'numerical': [1, 2, 3], 'categorical': ['a', 'b', 'a']}
df = pd.DataFrame(data)
print(df)

# Define transformers
numerical_features = ['numerical']
categorical_features = ['categorical']

# Create ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(sparse_output=False), categorical_features)])  # sparse=False for a dense array output

# Fit and transform
transformed_data = preprocessor.fit_transform(df)
print(transformed_data)
```

In this example:

    We import necessary libraries, including ColumnTransformer, StandardScaler, and OneHotEncoder.
    We create sample data with numerical and categorical columns.
    We define which columns should be treated as numerical and categorical.
    We create a ColumnTransformer object that will apply StandardScaler to numerical features and OneHotEncoder to categorical features.
    We use fit_transform to apply these transformations to our data and obtain the transformed data.
    To see the output, run the code.
